```python
import numpy as np
from sympy import Matrix, exp, I, sqrt, pi, symbols, cos, sin, tan
```

```python
# Define the symbols
theta1, theta2 = symbols('theta1 theta2')

# Define matrix B
B = Matrix([
    [1, I],
    [I, 1]
]) / sqrt(2)

# Define matrix R(theta1, theta2)
R = Matrix([
    [exp(I*theta1), 0],
    [0, exp(I*theta2)]
])

# Perform the matrix multiplication U = B * R * B
U = B * R * B

# Simplify the result
U.simplify()

# Display the result
print(U)
```

```python
def givens_generator(units, m, idx):
    theta1, theta2 = symbols(f'theta{idx} phi{idx}')
    theta2 = 0
    U = I * exp(I*(theta1 + theta2)/2) * Matrix([[sin((theta1-theta2)/2), cos((theta1-theta2)/2)], [cos((theta1-theta2)/2), -sin((theta1-theta2)/2)]])
    n = m + 1
    givens_rotation = Matrix(np.eye(units))
    givens_rotation[m, m] = U[0, 0]
    givens_rotation[m, n] = U[0, 1]
    givens_rotation[n, m] = U[1, 0]
    givens_rotation[n, n] = U[1, 1]
    return givens_rotation
```

```python
t1 = givens_generator(4, 0, 1)
t2 = givens_generator(4, 2, 2)
t3 = givens_generator(4, 1, 3)
t4 = givens_generator(4, 0, 4)
t5 = givens_generator(4, 2, 5)
t6 = givens_generator(4, 1, 6)
```
